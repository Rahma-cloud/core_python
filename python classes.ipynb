## Dictionary

```python
empty_dict = {}  
person = {
    "name": "Tomi",
    "age": 20,
    "city": "Ilorin"
}
numbers = {1: "one", 2: "Two", 3: "Three"}
mixed_keys = {"name": "Zainab", 42: "Answer", (1,2): "Tuple as key"}

print (person["name"]) # use key to access the values using []. NO Duplicate key, but we can have duplicate value
print(mixed_keys[42])
print(mixed_keys["name"])
print(numbers["four"]) # If a key doesn't exist, it'll raise a KeyError
```

```python
print(person["name"])
print (numbers.get("Tee", "Eat your food")) #get( helps to silence the keyError)
```

```python
#Adding a key-value pair to a dictionary
student = {
    "name": "Rahma",
    "age": 13
}
print (student)
student["city"] = "Ilorin"
print(student)

# Updating an existing Key's Value
student["age"] = 25
print(student)

#pop()- removes a key and returns its value; if the key does not exist, it raises a KeyError
person = {
    "name": 'Nike',
    "age": 34,
    "city": "Abuja",
    "language": "Hausa"
}
person.pop("age")
print(person)
```

```python
# popitem() - removes the last inserted key-value pair
schools = {
    "primary": "Iqra basic",
    "secondary": "Unilorin",
    "university": "Babcock"
}
print(schools)
schools.popitem()
print(schools)

#del - deletes a specific key
del schools["primary"]
print (schools)
```

```python
#DICTIONARY METHODS
# keys() - returns all keys
person = {"name": "Rayma", "age": 24, "city": "Ilorin"}
print(person)

print(person.keys())

#values() - returns all values
print(person.values())

#items() - returns key-value pairs as tuples
print (person.items())

# update() - merges two dictionaries
person = {"name": "Tofumi", "age": 1}
print(person)
new_data = {"city": "Ilorin", "age": 45}
person.update(new_data)
print(person)
#copy()
#len()
```

```python
# NESTED DICTIONARIES, i.e. a dictionary can have another dictionary
student = {
    "name": "Funmi",
    "info": {"age": 20, "hobbies": ["reading", "travelling", "cooking"]}
}
print(student['info']['hobbies'][1])
```

```python
a = 5 # Binary: 101
b = 3  # Binary: 011

print(a & b)  # 1 (Binary: 001)
print(a | b)   # 7 (Binary: 111)
print(a ^ b)   # 6 (Binary: 110)
print(~a)      # -6 (Binary: -(101+1))
print(a << 1)  # 10 (Binary: 1010)
print(a >> 1)  # 2 (Binary: 10)
```
