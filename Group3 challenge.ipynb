# 2 : Website Visitor Tracker - 10 points
You are tracking unique visitors to a website. Given the names of the visitors ['Abiola', 'Halimah', 'Zainab', 'Esther', 'Lola', 'Carina', 'Salamat', 'Funmi', 'Zainab', 'Jamilah', 'Lola', 'Hannah', 'Esther'].
Print all visitors
Print the unique visitors in alphabetical order

```python
visitors = ['Abiola', 'Halimah', 'Zainab', 'Esther', 'Lola', 'Carina', 'Salamat', 'Funmi', 'Zainab', 'Jamilah', 'Lola', 'Hannah', 'Esther']
print(visitors)
unique_visitors = set(visitors)
visitors_list = list(unique_visitors)
visitors_list.sort()
print(visitors_list)
```

# 4: Dictionary (Company Structure) - 20 points
You work for a company and you have the task of storing, updating and deleting the company's employee data in a dictionary (database). Each employee has a first name, last name, gender and role, and there are 4 employees.

a) Create a dictionary to store all 4 employees in a variable named 'webfala'.
b) Print every info about the second employee
b) Some time later, the 3rd employee was promoted to the CTO, update the role of the 3rd employee and then print the final data.


```python
webfala = {
    
    "employee_1":{
        "first_name": "Jane",
        "last_name": "Kola",
        "gender": "female",
        "role": "manager"
    },
    "employee_2": {
        "first_name": "Joy",
        "last_name": "Ali",
        "gender": "female",
        "role": "secretary"
    },
    "employee_3":{
        "first_name": "Ahmed",
        "last_name": "Muhammed",
        "gender": "male",
        "role": "CEO"
    },
    "employee_4": {
        "first_name": "Bisola",
        "last_name": "Femi",
        "gender": "female",
        "role": "CMO"
    },
} 
print(webfala.get("employee_2")) #getting all info of the second employee
webfala["employee_3"]["role"] = "CTO" #updating the new role
print(webfala.get("employee_3")) #printing the updated info
```
